SPRINT 1 (LEITURA E DIAGNÓSTICO)

APRESENTAÇÃO:

Nesta etapa foi baixado manualmente o arquivo CSV com os dados brutos (renomeado para "dados_brutos.csv") da plataforma Kaggle (https://www.kaggle.com/datasets/namespaiva/base-varejo/data). Os dados foram importados com a biblioteca Pandas para o Jupyter Notebook dentro do VS Code e feita uma verificação da estrutura com registro de valores ausentes, duplicatas, e inconsistências de texto.

```python
#Importando biblioteca Pandas
import pandas as pd

#Importando dados brutos
df = pd.read_csv('dados_brutos.csv', sep=';')

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Visualização rápida dos dados brutos (Primeiras 5 linhas)
print('Visualização rápida (Primeiras 5 linhas):')
print(df.head())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificando dimensões dos dados brutos
print('Dimensões (linhas, colunas):')
print(df.shape)

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificando dados válidos e tipos de dados nos dados brutos
print('Informações gerais (Dados válidos e tipo de dados):')
print(df.info())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificação e registro de valores ausentes nos dados brutos
print('Valores ausentes por coluna:')
print(df.isnull().sum())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificação e registro de LINHAS duplicadas nos dados brutos

print('Linhas totalmente duplicadas:', df.duplicated().sum())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificação e registro de VALORES duplicados nos dados brutos
print('Valores duplicados por coluna:')

for coluna in df.columns:
    duplicados = df[coluna].duplicated().sum()
    print(f'{coluna}: {duplicados}')

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificação e registro de inconsistências de texto no dados brutos

#Primeiro, verificar os registros de texto únicos
print('Registros únicos por coluna de texto:')

colunas_texto = df.select_dtypes(include=['object', 'string']).columns

print('Colunas textuais:')
print(colunas_texto.tolist())

for coluna in colunas_texto:
    print(f'Coluna: {coluna}')
    print(df[coluna].value_counts(dropna=False))

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Devido aos resultados da etapa anterior, é necessário contar a quantidade de "#N/D" em algumas colunas
print('Quantidade de #N/D na coluna PR_CAT:')
print((df[['PR_CAT']] == '#N/D').sum())
print('Quantidade de #N/D na coluna PR_NOME:')
print((df[['PR_NOME']] == '#N/D').sum())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Segundo, verificar as inconsistências de texto
print('Inconsistências por espaços (início/fim) e letras (maiusculas/minusculas) por coluna de texto:')

for coluna in colunas_texto:
    serie = df[coluna].dropna()
    
    espacos_extras = serie[serie != serie.str.strip()]
    
    valores = serie.unique()
    grupos = {}
    for valor in valores:
        chave = valor.strip().casefold()
        grupos.setdefault(chave, []).append(valor)
    variacoes = {chave: lista for chave, lista in grupos.items() if len(lista) > 1}

    print('Coluna: ' + coluna)

    if not espacos_extras.empty:
        print('Valores com espaços extras:')
        print(espacos_extras.unique())
    else:
        print('Boa notícia! Sem valores com espaços extras!')

    if variacoes:
        print('Possíveis variações textuais:')
        print(variacoes)
    else:
        print('Boa notícia! Sem variações textuais!')


```

CONCLUSÃO DO SPRINT 1

Os dados brutos foram importados com sucesso. A estrutura dos dados brutos é composta por 14 colunas com 830.000 linhas, sendo que 4 colunas são inválidas (sem nome e sem dados). Existem colunas com dados do tipo números inteiros e texto (as 4 colunas inválidas são do tipo números com casas decimais). Não existem dados faltantes nas 10 colunas válidas, porém, as colunas PR_CAT e PR_NOME possuem 3.650 registros de #N/D cada uma que devem ser tratados posteriormente. Existem 96.553 linhas totalmente duplicadas (que podem ser registros verdadeiros). Em cada coluna, também existem valores duplicados (o que não é necessariamente um erro). Não existem inconsistências nas colunas de texto devido a espaços no início e fim das palavras e por diferenças em letras maiúsculas e minúsculas nas palavras de mesma expressão. É também importante mencionar que a coluna de data está armazenada como texto e deverá ser convertida para o tipo "data". 

Apesar da maior parte do nome das colunas estar abreviada, é possível verificar que se tratam de dados de compra de produtos que são normalmente vendidos em conjunto em mercados (bebidas, alimentos, itens de limpeza, higiene, etc.), com registro de data de compra e gênero do(a) comprador(a). 

SRPINT 2 (LIMPEZA E PADRONIZAÇÃO)

APRESENTAÇÃO:

Nesta etapa foram corrigidos os problemas encontrados no SPRINT 1, justificadas as escolhas do tratamento dos problemas encontrados e feita uma comparação entre os dados brutos e os dados tratados nesta etapa (pré-processados).

```python
#Importando dados brutos com outro nome
df_t = pd.read_csv('dados_brutos.csv', sep=';')

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Exclusão das colunas sem dados
print('Exclusão das colunas sem dados!')
df_t = df_t.drop(columns=['Unnamed: 10'])
df_t = df_t.drop(columns=['Unnamed: 11'])
df_t = df_t.drop(columns=['Unnamed: 12'])
df_t = df_t.drop(columns=['Unnamed: 13'])

#Visualização rápida dos dados em tratamento (Primeiras 5 linhas)
print('Visualização rápida (Primeiras 5 linhas):')
print(df_t.head())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Substituição de registros "#N/D" por valores ausentes do Python
print('Exclusão das linhas com registros "#N/D"!')
df_t = df_t.replace('#N/D', pd.NA)
df_t = df_t.dropna(subset=['PR_CAT', 'PR_NOME'])

#Checagem da exclusão de linhas com valores ausentes
print('Dimensões (linhas, colunas):')
print(df_t.shape)
print('Valores ausentes por coluna:')
print(df_t.isnull().sum())
print('Quantidade de #N/D na coluna PR_CAT:')
print((df_t[['PR_CAT']] == '#N/D').sum())
print('Quantidade de #N/D na coluna PR_NOME:')
print((df_t[['PR_NOME']] == '#N/D').sum())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Converter os registros das data armazenados como texto para formato de data
print('Conversão das datas armazenadas como texto para formato de data!')
df_t['DATA'] = pd.to_datetime(df_t['DATA'], format='%d/%m/%Y')

#Checagem rápida do tratamento
print('Formato dos dados da coluna DATA:')
print(df_t['DATA'].dtype)
print('Visualização rápida (Primeiras 5 linhas):')
print(df_t.head())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Comparação entre os dados brutos e os dados tratados
print('Dados brutos (linhas, colunas):', df.shape, '| Dados tratados:', df_t.shape)
print('Dados brutos (informações básicas):')
print(df.info())
print('Dados tratados (informações básicas):')
print(df_t.info())

```

a) 4 colunas sem nome e sem dados;

b) Os registros de "#N/D" (erros referentes a dados ausentes) nas colunas "PR_CAT" e "PR_NOME";

c) Os registros de data armazenados como texto. 

Como tratamento, optou-se por ecxluir as 4 colunas sem nome e sem dados, exluir as linhas com registro de "#N/D" nas colunas "PR_CAT" e "PR_NOME" (0,44% das linhas), e converter os registros de data para o formato correto de data.

Em relação às linhas totalmente duplicadas, considerou-se ser normal que em um comércio varejista do tipo mercado, ocorra a venda dos mesmos tipos de produto, pelo mesmo cliente, na mesma compra. Em relação as duplicatas na mesma coluna, no referido tipo de atividade econômica é normal várias compras dos mesmos produtos no mesmo dia e em dias diferentes, e por todos os tipos de gêneros registrados na base de dados.

É também importante mencionar que não foram encontrados inconsistências de texto, e por isso não foi necessário fazer o tratamento/padronização destes dados.
