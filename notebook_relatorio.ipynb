SPRINT 1 (LEITURA E DIAGNÓSTICO)

APRESENTAÇÃO:

Nesta etapa foi baixado manualmente o arquivo CSV com os dados brutos (renomeado para "dados_brutos.csv") da plataforma Kaggle (https://www.kaggle.com/datasets/namespaiva/base-varejo/data). Os dados foram importados com a biblioteca Pandas para o Jupyter Notebook dentro do VS Code e feita uma verificação da estrutura com registro de valores ausentes, duplicatas, e inconsistências de texto.

```python
#Importando biblioteca Pandas
import pandas as pd

#Importando dados brutos
df = pd.read_csv('dados_brutos.csv', sep=';')

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Visualização rápida dos dados brutos (Primeiras 5 linhas)
print('Visualização rápida (Primeiras 5 linhas):')
print(df.head())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificando dimensões dos dados brutos
print('Dimensões (linhas, colunas):')
print(df.shape)

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificando dados válidos e tipos de dados nos dados brutos
print('Informações gerais (Dados válidos e tipo de dados):')
df.info()

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificação e registro de valores ausentes nos dados brutos
print('Valores ausentes por coluna:')
print(df.isnull().sum())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificação e registro de LINHAS duplicadas nos dados brutos

print('Linhas totalmente duplicadas:', df.duplicated().sum())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificação e registro de VALORES duplicados nos dados brutos
print('Valores duplicados por coluna:')

for coluna in df.columns:
    duplicados = df[coluna].duplicated().sum()
    print(f'{coluna}: {duplicados}')

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Verificação e registro de inconsistências de texto no dados brutos

#Primeiro, verificar os registros de texto únicos
print('Registros únicos por coluna de texto:')

colunas_texto = df.select_dtypes(include=['object', 'string']).columns

print('Colunas textuais:')
print(colunas_texto.tolist())

for coluna in colunas_texto:
    print(f'Coluna: {coluna}')
    print(df[coluna].value_counts(dropna=False))

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Devido aos resultados da etapa anterior, é necessário contar a quantidade de "#N/D" em algumas colunas
print('Quantidade de #N/D na coluna PR_CAT:')
print((df[['PR_CAT']] == '#N/D').sum())
print('Quantidade de #N/D na coluna PR_NOME:')
print((df[['PR_NOME']] == '#N/D').sum())

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Segundo, verificar as inconsistências de texto
print('Inconsistências por espaços (início/fim) e letras (maiusculas/minusculas) por coluna de texto:')

for coluna in colunas_texto:
    serie = df[coluna].dropna()
    
    espacos_extras = serie[serie != serie.str.strip()]
    
    valores = serie.unique()
    grupos = {}
    for valor in valores:
        chave = valor.strip().casefold()
        grupos.setdefault(chave, []).append(valor)
    variacoes = {chave: lista for chave, lista in grupos.items() if len(lista) > 1}

    print('Coluna: ' + coluna)

    if not espacos_extras.empty:
        print('Valores com espaços extras:')
        print(espacos_extras.unique())
    else:
        print('Boa notícia! Sem valores com espaços extras!')

    if variacoes:
        print('Possíveis variações textuais:')
        print(variacoes)
    else:
        print('Boa notícia! Sem variações textuais!')

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')
```

CONCLUSÃO DO SPRINT 1

Os dados brutos foram importados com sucesso. A estrutura dos dados brutos é composta por 14 colunas com 830.000 linhas, sendo que 4 colunas são inválidas (sem nome e sem dados). Existem colunas com dados do tipo números inteiros e texto (as 4 colunas inválidas são do tipo números com casas decimais). Não existem dados faltantes nas 10 colunas válidas, porém, as colunas PR_CAT e PR_NOME possuem 3.650 registros de #N/D cada uma que devem ser tratados posteriormente. Existem 96.553 linhas totalmente duplicadas (que podem ser registros verdadeiros). Em cada coluna, também existem valores duplicados (o que não é necessariamente um erro). Não existem inconsistências nas colunas de texto devido a espaços no início e fim das palavras e por diferenças em letras maiúsculas e minúsculas nas palavras de mesma expressão. É também importante mencionar que a coluna de data está armazenada como texto e deverá ser convertida para o tipo "data". 

Apesar da maior parte do nome das colunas estar abreviada, é possível verificar que se tratam de dados de compra de produtos que são normalmente vendidos em conjunto em mercados (bebidas, alimentos, itens de limpeza, higiene, etc.), com registro de data de compra e gênero do(a) comprador(a).

SPRINT 4 (ANÁLISE)

APRESENTAÇÃO:

Nesta etapa são gerados dois agrupamentos e posteriormente são calculadas as estatísticas básicas destes agrupamentos. Os agrupamentos foram feitos para as vendas por categoria de produtos e por gênero dos clientes. Devido a limitação nos dados quantitativos (colunas com nome abreviado, de difícil interpretação), optou-se por usar o tipo de produto com representação de vendas, ou seja, cada produto representa uma venda. Também é aplicado um filtro, feito um agrupamento e calculadas suas estatísticas básicas. 

```python
#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Primeiro agrupamento e suas estatísticas descritivas: Vendas por categoria de produto

print('Estatísticas das vendas por categoria de produtos:')
#Contagem de produtos por categoria (cada produto representa uma venda)
cat_cont_1 = df_t.groupby('PR_CAT')['PR_NOME'].value_counts()
#Cálculo das estatísticas básicas (contagem, mínimo, máximo e média)
categoria_1 = cat_cont_1.groupby(level=0).describe()[['count', 'min', 'max', 'mean']]
#Cálculo da soma de produtos por categoria
categoria_1['sum'] = cat_cont_1.groupby(level=0).sum()
#Contagem, mínimo e máximo como número inteiros (a casa decimal é sempre zero nessas estatísticas)
categoria_1[['count','min','max']] = categoria_1[['count', 'min', 'max']].astype(int)
#Média com duas casas decimais
categoria_1['mean'] = categoria_1['mean'].round(2)
print(categoria_1)

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Segundo agrupamento e suas estatísticas descritivas: Vendas por gênero de clientes

print('Estatísticas de vendas por gênero de clientes:')
#Contagem de produtos por gênero (cada produto representa uma venda)
cat_cont_2 = df_t.groupby('CL_GENERO')['PR_NOME'].value_counts()
#Cálculo das estatísticas básicas (contagem, mínimo, máximo e média)
categoria_2 = cat_cont_2.groupby(level=0).describe()[['count', 'min', 'max', 'mean']]
#Cálculo da soma de produtos por categoria
categoria_2['sum'] = cat_cont_2.groupby(level=0).sum()
#Contagem, mínimo e máximo como número inteiros (a casa decimal é sempre zero nessas estatísticas)
categoria_2[['count','min','max']] = categoria_2[['count', 'min', 'max']].astype(int)
#Média com duas casas decimais
categoria_2['mean'] = categoria_2['mean'].round(2)
print(categoria_2)

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Filtrar apenas clientes do gênero feminino (F)
df_f = df_t[df_t['CL_GENERO'] == 'F']

print('Filtro aplicado para o gênero feminino')
print('Linhas antes do filtro:', df_t.shape[0], '| Linhas após o filtro:', df_f.shape[0])

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')

#Agrupamento e estatísticas descritivas: Vendas por categoria de produto (apenas mulheres)
print('Estatísticas das vendas por categoria de produtos (gênero F):')
#Contagem de produtos por mulheres e por categoria (cada produto representa uma venda)
cat_cont_f = df_f.groupby('PR_CAT')['PR_NOME'].value_counts()
#Cálculo das estatísticas básicas (contagem, mínimo, máximo e média)
categoria_f = cat_cont_f.groupby(level=0).describe()[['count', 'min', 'max', 'mean']]
#Soma de produtos por categoria
categoria_f['sum'] = cat_cont_f.groupby(level=0).sum()
#Contagem, mínimo e máximo como números inteiros
categoria_f[['count','min','max']] = categoria_f[['count', 'min', 'max']].astype(int)
#Média com duas casas decimais
categoria_f['mean'] = categoria_f['mean'].round(2)
print(categoria_f)

#Espaçamento de organização da apresentação dos resultados
print('\n' + '-' * 50 + '\n')
```

CONCLUSÃO DO SPRINT 4

Com base em dois agrupamentos, um baseado na categoria de produtos e outro baseado no gênero dos clientes, foram analisadas as quantidades de vendas, considerando que cada tipo de produto, que possui uma linha específica na base de dados, representa uma venda. 

A categoria de produtos com mais tipos de produtos são os ALIMENTOS e os com menos tipos de produtos são os ACESSÓRIOS. Os tipos de produtos com menos e mais vendas são da categoria ALIMENTOS. Em média, a categoria que possui mais vendas por tipo de produtos são os ACESSÓRIOS (que possui só dois tipos de produtos). No entanto, a venda total de produtos por categoria é expressivamente maior na categoria ALIMENTOS, enquanto a menor venda total ocorre na categoria ACESSÓRIOS. 

Também foi verificado que as mulheres compram maior quantidade de tipos de produtos do que os homens, e o padrão de compras por categoria de produtos é muito similar ao apresentado anteriormente, se diferenciando apenas no valor mínimo por tipo de produto, que agora passa a ser da categoria PET.
